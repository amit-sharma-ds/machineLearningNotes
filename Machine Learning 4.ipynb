## LOGISTIC REGRESSION

* Logistic regression is one of the most popular Machine Learning algorithms, 
which comes under the Superviser Learning technique.

* It is used for predicting the categorical dependent variable using a given set of independent variables.

* Therefore, the outcome must be a categorical or discrete value. It can be either Yes or No, 
0 or 1, true of False, etc. but instead of giving the exact value as 0 and 1, it gives the probablistic
values which lie between 0 and 1.

## TYPE OF LOGISTIC REGRESSION

On the basis of the categories, Logistic Regression can be classifier into three types:

* <span style="color:red">Binomial:</span> In binomial Logistic regression, there can be only two possible types of the 
dependent variables, such as 0 or 1, Pass or Fail, etc.

* <span style="color:red">Multinomial:</span> In multinomial Logistic regression, there can be 3 or more possible unordered
types of the dependent variable, such as "cat", "dogs", or "sheep".

* <span style="color:red">Ordinal:</span> In ordinal Logistic regression, there can be 3 or more possible 
ordered types of dependent variables, such as "low", "Medium", or "High".

### In short: Important

We can apply logistic regression only when our data is `linearly separable`.
    If data is non-linear separable data then we can not apply logistic regression.

So, first we will check the nature of the data. If data is linearly separable
        then we can apply logistic regression other wise not.

```python
# Explanation types
# --------------------------

# Lr categories in 3 types.

# 1. Binomial: In binomial category the output is divided into 2 parts.
#     binary means 2. 
#     means (true, false), (yes, no) ya phir (cat and dog).

#     This is binary class analysis.

# 2. Multinomial: In multinomial logistic analysis in which our data is separable into more than 2 parts.
#         ex: cow, dog, buffalo. or mouse, pen, remote.
#         In this way the data is divided into multiple parts that is called of multi-nomial category.

# 3. Ordinal: here also the data is divided into more than 2 parts.
#     but the classified data should be in a order.
#     ex: m, l, xl or low, medium, high

#     so, basically here data is separable but in a order.
```

```python
# // prediction line using linespace

from sklearn.linear_model import LogisticRegression
import numpy as np
import matplotlib.pyplot as plt

# Fit logistic regression
X = dataset[["Age"]]
y = dataset["Purchased"]
model = LogisticRegression()
model.fit(X, y)

# Generate prediction line
X_test = np.linspace(X.min(), X.max(), 300).reshape(-1, 1)
y_pred = model.predict_proba(X_test)[:, 1]

plt.figure(figsize=(4,3))
plt.scatter(X, y, color='blue')
plt.plot(X_test, y_pred, color='red')
plt.xlabel("Age")
plt.ylabel("Purchased")
plt.show()
```

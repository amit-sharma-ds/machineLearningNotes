## i) What is Machine Learning?

Machine Learning is the science of getting computers to **learn and act like humans**, and **improve their learning over time in an autonomous fashion** by feeding them data and information in the form of observations and real-world interactions.

With the help of machine learning, we can perform:

- **Future predictions**
- **Recommendation generation**

### 🔍 Real-Life Examples:

1. **YouTube** creates a playlist of mixed videos based on your interests.
2. **Flipkart/Amazon** shows **"Frequently Bought Together"** suggestions.
3. Email services detect **Spam vs Inbox** (Ham) mails automatically.

## ii) How Does Machine Learning Work?

A Machine Learning system:

1. **Learns from historical data**
2. **Builds prediction models**

## iv) Classification of Machine Learning

1. **Supervised Learning (SL)**
   - Used for **future prediction or recommendations**
   - **Input and output are given**
   - Example: Predicting house prices, email spam detection

2. **Unsupervised Learning (UL)**
   - Used for **grouping/clustering**
   - No predefined labels (only input is given)
   - Example: Grouping different types of fruits

3. **Reinforcement Learning (RL)**
   - Learning through **interactions and feedback**
   - Example: Games like Ludo King, Chess, where the computer learns strategy

> 🔄 Each ML type works differently in different stages and application areas.


## v) Advantages of Machine Learning

1. Easily identifies **trends and patterns**
2. **No human intervention** needed (automation)
3. **Continuous improvement** over time
4. Handles **multi-dimensional and multi-variety data**
5. Wide range of **applications**

```python
si = SimpleImputer(strategy="mean") # calling the class/creating object of class
ar = si.fit_transform(dataset[['ApplicantIncome','CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term',
       'Credit_History']])


# si varible me hmare SimpleImputer class ready gya hai
# now si ko fit_transform krenge taki mean apply ho jaye null values pe.



# shift tab inside bracket to show documentation

# Init signature:
# SimpleImputer(
#     *,
#     missing_values=nan,
#     strategy='mean',
#     fill_value=None, (jisse fill krna hai)
#     copy=True,
#     add_indicator=False,
#     keep_empty_features=False,
# )

# we can see many hyperparameters inside simpleimputer class

# we can fill with constant value also
```

```python
# converting above array into dataframe
new_dataset = pd.DataFrame(ar,columns=dataset.select_dtypes(include=["float64","int64"]).columns)
```

```python
# checking missing value
new_dataset.isnull().sum()

# now no missing value is present
```

# OHE VS LE VS OE

___
**<u>One-Hot Encoding (OHE)</u>**

* Definition: Converts each unique category into a separate binary column (0 or 1).
* Representation: Each category becomes a new column in the dataset.

<u>Ohe</u>
| Category | Red | Blue | Green |
|----------|-----|------|-------|
| Red      |  1  |  0   |   0   |
| Blue     |  0  |  1   |   0   |
| Green    |  0  |  0   |   1   |

    
* Use cases: For unordered input features (Nominal data).
* Pros: 
    
    * Avoids implying an ordinal relationship between categories.
    * Works well with ML algorithms that don’t assume order (e.g., Logistic Regression, Decision Trees).

* Cons: Increases dimensionality; creates sparse matrices for high-cardinality features.

___
**<u>Label Encoding</u>**

* Definition: Assigns a unique numeric value to each category.
* Representation: Converts categories into a single column with numeric values.
Example:
| Category | Label |
|----------|-------|
| Red      |   0   |
| Blue     |   1   |
| Green    |   2   |

* Use cases: Mostly for target/output labels (classification problems).
* Pros:
    * Simple and memory-efficient.
    * Easy to implement for algorithms that can handle categorical data directly.
* Cons:
    * Implies an ordinal relationship (e.g., 0 < 1 < 2) which might mislead models for input features.



___
**<u>Ordinal Encoding</u>**

* Definition: Maps categories to numeric values based on their natural order.
* Representation: Categories are assigned numbers in increasing order.
Example:
| Category | Ordinal |
|----------|---------|
| Small    |   0     |
| Medium   |   1     |
| Large    |   2     |

* Use cases: For both input features and target labels when categories have a clear, meaningful order.
* Pros: 
    * Captures the natural relationship between categories.
    * Memory-efficient and easy to use with ordered data.
* Cons:
    * Not suitable for unordered categories.
    * Implies order even if it doesn’t exist, leading to incorrect results for nominal data.

# <span style = "color:skyblue">Standardization Example (my example) </span>

Let's see step by step how standardization makes **mean = 0** and **variance = 1**.

## 🧮 Given Data
x = [1, 2, 3, 4, 5]

## Step 1️⃣: Find the Mean (p)

The formula for mean:

$$
p = \frac{1 + 2 + 3 + 4 + 5}{5} = \frac{15}{5} = 3
$$

👉 **Mean (p) = 3**

---

## Step 2️⃣: Find the Standard Deviation (q)

The formula for standard deviation:

$$
q = \sqrt{\frac{(1-3)^2 + (2-3)^2 + (3-3)^2 + (4-3)^2 + (5-3)^2}{5}}
$$

$$
q = \sqrt{\frac{4 + 1 + 0 + 1 + 4}{5}} = \sqrt{2} \approx 1.414
$$

👉 **Standard Deviation (q) ≈ 1.414**

---

## Step 3️⃣: Apply the Standardization Formula

The standardization formula:

$$
X_{\text{new}} = \frac{X_i - p}{q}
$$

| xᵢ | Formula          | Xₙₑw Value |
|----|-----------------|------------|
| 1  | (1 - 3)/1.414   | -1.414     |
| 2  | (2 - 3)/1.414   | -0.707     |
| 3  | (3 - 3)/1.414   | 0          |
| 4  | (4 - 3)/1.414   | 0.707      |
| 5  | (5 - 3)/1.414   | 1.414      |

---

## Step 4️⃣: Check Mean and Variance

**Mean of Xₙₑw:**

$$
\bar{X}_{\text{new}} = \frac{-1.414 + (-0.707) + 0 + 0.707 + 1.414}{5} = 0
$$

✅ **Mean ≈ 0**

**Variance of Xₙₑw:**

$$
\text{Var}(X_{\text{new}}) = \frac{(-1.414)^2 + (-0.707)^2 + 0^2 + 0.707^2 + 1.414^2}{5} = 1
$$

✅ **Variance = 1**

---

## 🎯 Conclusion

After standardization:

$$
\text{Mean} = 0, \quad \text{Variance} = 1
$$

This is why **standardization re-scales the data** so that it has **zero mean** and **unit variance**.

```python
# we will use IQR method here to remove outlier.

q1 = dataset["CoapplicantIncome"].quantile(0.25)
q3 = dataset["CoapplicantIncome"].quantile(0.75)
iqr = q3-q1
```

```python
# Note: 
########## we will use log1p

# * we can use log base 10 and log base 2 also.
    # but if we use logbase10:  and agr zero aa gya hai kisi condition ke andr.
    # then agr kisi condition ke andr 0 aa rkha hai then log 0 ki value work ni krte hai.
    # to agr hm log(0), log base 10, log 1c to agr wha pe hm 0 value dete hai.
    # to ye work nhi kr pata hai.


# * so we will use log1p
    # means (data - 1) means agr hm apne data me se 1 ko minus krenge to
    # hmare pass ek normal value aa jayegi.

# * to kbhi hmare dataset ke in case kbhi zero(0) bhi present hai.
    # and we want that hmara data working krte rhe. and error na show ho.
    # then we will use 1p for it.


####   understand why log1p not log base 10?
# ---------------------------------------------------

# * we don't know ki hmara data kya ane wala hai.
    # hmara data 0 se start hoke, kisi bhi data point pe ja skta hai.
    # means kisi bhi no. pe ja skta hai, infinity pe bhi ja skta hai.

# * to basically hmara data 0 se start hota hai. 
    # ab 0 se start hota hai, to agr hm log base 0, lete hai.
    # ya base ke 10 lete hai, ya 10 base ka use krte hai, tb bhi hmara data
    # infinity pe chla jayega.
    # ye e base ke sath bhi agr hm 0 ka use krte hai, tb bhi hmara data infinity bn jayega.

# * to in sb cases me hme infinity milne wala hai.
    # and infinity is basically a undefined thing.
    # and ye hmare dataset ko khrab kr skta hai.
    # so isiliye we use (p-1)  means (data-1) ka use krte hua kam me lete hai.


### practical on calculator
# ---------------------------------
# log(0) -> Invalid input (means infinity)
# In(0) -> Invalid input (means infinity)


# * In short:
# -----------------
    # agr hm log 10 le, ya log base e lenge.
    # agr hmara data column me zero value ata hai.
    # to hmara data infinity ke trh chala jata hai.
    # and it does not give any desired output.

# * to isilye hm yha pr 1p ka use kr rhe hai.
    # 1p (p-1) ka output de rha hai.
    # mtlb hmara data point me se 1 ko less kiye ja rha hai/
    # taki hmara result shi bn skte
    
    
# * and hme yha pe pta hai ki hmara "CoapplicantIncome" ke andr 
    # data 0 hai, but hme kbhi ni pta hota hai ki column me data 0 hai.
    # ya ni, then we can use 1p
```

## Forward Elimination



image at 3.13.00

Forward Stepwise Selection works by iteratively adding features that improve the model's performance the most. Here's how you handle this process:

1. First Step:

* Evaluate the accuracy of the model using each single feature individually. In your diagram, the best-performing single feature is Feature 3 (with 75% accuracy).
* Select Feature 3 as the first feature to include in the subset.

2.Second Step:

* Combine Feature 3 with each remaining feature one at a time (e.g., Feature 3 + Feature 1, Feature 3 + Feature 2, and so on).
* Calculate the accuracy of the model for each of these combinations.
* Choose the combination that results in the highest accuracy. For example, in the diagram, Feature 3 + Feature 1 gives the highest accuracy (let's assume it's 75% again).

3. Third Step:

* Repeat the process by combining the already selected subset (Feature 3 + Feature 1) with the remaining features one at a time.
* Evaluate the model's accuracy for each combination (e.g., Feature 3 + Feature 1 + Feature 2, Feature 3 + Feature 1 + Feature 4, etc.).
* Select the subset with the highest accuracy.

Important:
*You only need to evaluate combinations that include the previously selected features (e.g., start with Feature 3, then Feature 3 + Feature X, etc.). This avoids unnecessary computation for combinations that do not include the best features from the earlier steps.

Note: feature 3 ki accuracy highest hai to next bonding highest feature ke sath hi bnega

<u>In Forward Stepwise Selection, you always select the feature combination that provides the highest accuracy at each step. If, for example, the accuracy of a combination (like Feature 3 + Feature 1) is 73% (which is lower than the previous highest accuracy of 75%), you would not choose that combination.<u/>

<u>Here's the general rule:</u>

* At each step, you evaluate all possible feature combinations (starting from individual features to combinations) and select the one that gives the highest accuracy.
* If the accuracy for a new combination is lower than the previous highest, you do not select it. You keep the previous combination.
* The process continues by adding the feature that maximizes the accuracy for the next subset.
So, if the 73% accuracy for Feature 3 + Feature 1 is lower than 75% (or any previous best), you would skip it and move forward with the combination that gives you the best result.

**<u>In summary:</u>**
* Always check the highest accuracy.

* If the accuracy is lower than the previous step, do not select that feature combination.

```python
# In short (FORWARD ELIMINATION)
# ----------------------------------------

# i) forward elimination me hm all features ko lete hai and unki accuracy check krte hai.
# jiski accuracy jada hoti hai as feature 3 usko select krte hai.

# ii) now feature 3 ke sath sbse pair/bonding bnayega.
# jis combination ka accuracy highest hoga usko select krenge.

# but note: ye highest to hai but hme dekhenge ki kya ye previous wale step se bhi highest hai.
#     if yes then we will move ahead.
#     other wise the feature selection process we will stop.
    
# iii) feature 3 and 1 has highest from all the pairs and also from the previous which was 75%.
# so now we will create 3-3 combinations one by one.
```

### 2. Soft Margin

```python
hmare pass esi kafi situation aa jati hai, 
for ex: ydi hmare pass, kuch data points is treh se ata hai,
jaise (see in the image) .
    hmare pass kuch data points is formation me aa jaye,
```

```python
1. Hard margin: in the figure 1. 
    this is hard margin. isme to hm easily work kr skte hai.

2. Soft margin: see the 2nd figure.
    let's suppose hmare pass jo data set hai,
wo kuch is treh se hai, to 

us wqt hmko soft margin ki jrurt pdegi.
    bcoz wha pe hm data point ko perfectly separable nhi kr payenga.

ye shi treke se separable nhi payenga.
    bcoz the data points are mix up.

to agr hm poistive and negative hyperplain ko draw kre.
        aur then hm apne line ko separable bnaye.

To distace d to kafi maximum hai, but iske bich me kafi mis classification hai,
```

## KERNEL FUNCTIONS 

* kernel functions play a crucial role in transforming input data into a higher-dimensional space.

* The primary purpose of kernel functions is to allow SVMs to handle non-linearly
separable data by implicitly mapping the input data into a higher-dimensional feature
space where linear separation may be more feeasible.

* This transformation is done without explicitly calculating the coordinate data
points in that higher-dimensional space.

```python
Kernel converts one dimensional data into high dimensional space.
So, that we can do that's splitting easily.
```

# 2. Elbow Method:

Elbow Method:

* The Elbow method is one of the most popular ways to find the optimal number of 
clusters.

* This method uses the concept of WCSS value. WCSS stands for Within Cluster Sum
of Squares, which defines the total variations within a cluster.

$$
WCSS = \sum_{p_i in Cluster_1} distance(p_i C_1)^2
     + \sum_{p_i in Cluster_2} distance(p_i C_2)^2
     + \sum_{p_i in Cluster_3} distance(p_i C_3)^2
$$



img at v2.3.57.00

//my formula
$$
\text{WCSS} = \sum_{j=1}^{k} \sum_{P_i \in \text{cluster}_j} \text{distance}(P_i, C_j)^2
$$

```python
### Explanation 

i. Elbow Method/ WCSS method: iske help se hm, best no. groups ko find out krte hai.
    mtlb if we have a dataset. Then usme kitne no. of cluster bnege ye hme WCSS method btata hai.

ii. Within a Cluster Sum of Squares: isko hm error bol skte hai.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Unlabelled
dataset = sns.load_dataset("iris").drop(columns="species")
```

```python
dataset.head(3)

## It has no output. Now, we will create clusters of this dataset.
```

```python
## To check the data is linearly separable of not

# i. Before making the clusters, we will do analysis in it's graph.

# ii. ki kis-kis treh ke graphs hme dikh rhe hai, means what relationship seeming between them.
# taki hm yha pe kam le ske. 

# iii. Bcoz The k-mean clustering is always and always used for linear separable.

# iv. We have to use K-mean clustering only when we have linearly separable data.   <---- very important
#     If non-linearly separable data then we have not to use K-mean clustering.

# v. for non-linear, we have more algorithms which are capable to make clusters.
```

```python
## creating graph. To check the data is linearly separable of not
sns.pairplot(data=dataset)
plt.show()

## observation

# i. see sbse niche ke 3 graph.
#     petal width vs sepal length
#     petal width vs sepal width
#     petal width vs petal length

# here, hme bs 2 clusters dikh rhe hai, but we know there are 3 no. of clusters.
# setos, virginica, vergicolor.

# we have to find these 3 no. of clusters.

# K-Means cluster will helps us to find these clusters.
```

```python
### Note:  Important

# i. In k-mean clustering we do not split data into training and testing.
#     bcoz here data is unlabelled data.
# aur unlabelled data ko training aur testing ke andr splitting krke hm krenge bhi kya.

# ii. means traing aur testing use when we are working in SML.

# iii. In SML first we train data. after train we test it. that they are working right or not.

# iv. But in UML here the data is unlabelled. To basically yha pe training nhi hoti.
#     Yha pe basically patterns ke hisab se groups ke andr divide kiya jata hai.
#     To basically here training aur testing ke andr splitting nhi hoga.
#     here we will find no. of clusters.
#     That how many best no. of clusters are formed.


### In short:

# # Train test split use in SML not in UML bcoz of unlabelled data.
# Here, we divided the data into groups on the bases of patterns.
```

```python
# ### To make best no. of clusters

# we will use scikit-learn library.
# Bcoz muje yha need hogi WCSS ki. Here, we have to find Within a clusters sum square.

# aur uski help se we will find best no. of clusters.
# Means with the help of Elbow method we are finding.


# ### In short

# We will find best no. of clusters using Elbow/ WCSS method.
```

```python
from sklearn.cluster import KMeans
```

```python
## finding wcss value

wcss = []

for i in range(2,21):
    km = KMeans( n_clusters=i,init='k-means++',)
    km.fit(dataset) ## <-- fit means we have to transfer/ pass our data. We have data in the form of dataset.
    wcss.append(km.inertia_)
    
# i. No idea how many no. of clusters: Abhi kitne no. of clusters yha pe bnne wale hai, we don't have any idea.
# so hm ek loop chlayenge. uske andr hm 2 clusters lgake, 20 clusters tk bnayenge.

# ii. Loop: Mtlb we will create 2 to 20 clusters. In which we will see our best no. of clusters.

# iii. Understanding it's hyperparameters (shift + tab)
    # KMeans(
    #     n_clusters=8,  <-- no. of clusters
    #     *,
    #     init='k-means++',  <-- initiate 
    
    # initiate ke andr agr k-means++ nhi hai, then we have to give this.
    # bcoz it's help us to find the best no. of clusters.

# iv. km.inertia: yhi wcss ki value hota hai, which gives us wcss value through which we will design elbow graph.

# v. wcss.append -> wcss is a list here.

# vi. after running this code. our value will be trained.

# vii. Note: if we will get warning in the output, then we can ignore that
    # bcoz we are using and python version kafi bar wo iske sath supported nhi hota hai,
    # to uske regarding yha pe kuch warning di hoti hai, so, we can ignore it.
    
    # But if error comes then we can't. that's a not ignorable.
    # but here we have only warning. to ek bar ke liye we can ignore them.
```

```python
## Designing graph b/w wcss value vs no. of clusters

plt.figure(figsize=(10,5))
plt.plot([i for i in range(2,21)],wcss,marker="o")
plt.xlabel("no of cluster")
plt.xticks([i for i in range(2,21)])
plt.ylabel("WCSS")
plt.grid(axis = "x")
plt.show()

# plt.plot(x,y)
# x axis -> cluster value, [i for i in range(2,21)]
# y axis -> wcss value

# marker="o" <-- ye hr data point ke upr, means hr cut ke upr gives dot.
# it gives us markers point.

# grids: shift + tab. we can apply grid according to the axis.
```

```python
# ## Elbow point in above graph

# i. we have to find a elbow point jha pe ye data settle hone wala hai.

# ii. here settle point is 3. onwards 3 graph downwards hota chla ja rha hai.

# iii. Elbow point is 3 it means 3 no. clusters yha pe desing hone wala hai.

# iv. first point elbow point mil rha hai. aur elbow point ke badh me hmara hand sidha ho jata hai.
# here, we can also observe this thing.

# v. ye no. of clusters hme 3 design de rha hai.-> so that best no. of cluster is 3.
    

# vi. waise yha pe 3 aur 4 me confusion rhega. but 3 aur 4 se hm jo bhi point lena chahiye hm le skte hai. 

# vii. so, 3 bhi hai, 4 bhi hai, but according to our dataset. 3 no. of clusters bnne chahiye.
#     so, we are designing 3 no. of clusters.

# ## In short
#     here the elbow point is 3. It means 3 no. of clusters we will design.
#     elbow point ke badh hmare hand straing ho jate hai, we can observe this in the above graph as well.
```

```python
# designing 3 no. of clusters.
knn = KMeans(n_clusters=3)
dataset["Predict"] = knn.fit_predict(dataset)

## predicted value saved in predict column
```

```python
# //
dataset

# we can see predict column
```

```python
sns.pairplot(data=dataset,hue="Predict")
plt.savefig("Predict.jpg")
plt.show()

# here, we can see the clusters. prediction is very good here.
```

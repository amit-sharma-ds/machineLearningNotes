```python
Lets' suppose, hmare koi nya data point aa rha hai. see the yellow point.

to ye new data point, apne njdik ke pdosiyo ko dhundhta hai.

Now, ye njdik ke pdosiyo ko kaise dhunda hai.

ye data points kya krta hai, apne jitne bhi data points hai,
mtlb iske aas pass jitne bhi data points hai, un sbki distance ko calculate krta hai.

bcoz hme njdik ke pdosi nikalna hai, mtlb nearest nikalna hai, 
iske liye hme sbhi data points se , hme distance calculate krna pdega.

Un distance ke andr, hme filter out krna pdega ki.

usme se jo sbse shortest distance hai, 

shortest means: jo sbse njdik ke distance hai, wo nikalni pdegi.

Then, usme se bhi top 5 njdik ke distance ko find out krna pdega.
```

```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset = pd.read_csv(r"C:\Users\ATVIK SHARMA\OneDrive\DataSets\Social_Network_Ads.csv")
```

```python
# //
dataset.head(3)
```

```python
# //
dataset.drop(columns=["User ID","Gender"],inplace=True)
```

```python
dataset.head(3)
```

```python
sns.scatterplot(x="Age",y="EstimatedSalary",data=dataset,hue="Purchased")
plt.show()
```

```python
## Checking null values

dataset.isnull().sum()
```

```python
# Differentiating data into input and output
# input and output means separate data into dependent and independent

# and then scaling krte hai.
# -----------------------------------------------

x = dataset.iloc[:,:-1]
y = dataset["Purchased"]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
# //
# sc = StandardScaler()
# sc.fit(x)
# sc.transform(x)

# Why scaling
# ----------------

# ye hmne pehle bhi kiya tha scaling. is dataset me.
# Scaling isliye kiya hai because Age and EstimatedSalary ki values size mein bohot different hain, 
# aur bina scaling ke model unhe equal importance nahi de paata.

# ---------------------------------------------------------------
# is data ko hm dataset me wapas place kr dete hai.
#     see below the complete code.
```

```python
sc = StandardScaler()
sc.fit(x)
x = pd.DataFrame(sc.transform(x),columns=x.columns)
```

```python
# //
x

# here, we can see that x axis ka dataset scaling hoke aa chuka hai.
```

```python
# splitting the dataset into train and test.
#     so that we can check that how well this model is working.

from sklearn.model_selection import train_test_split

# //Read> scikit learn ke andr jake model_selection ko call krenge, and then hm call krenge train test splitting ko
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,random_state=42,test_size=0.2)

# our dataset has been splitted into 4 parts
```

```python
## Model Building
```

```python
from sklearn.neighbors import KNeighborsClassifier


# Explanation
# scikit learn ke andr jayenge neighbor ko call krenge,
# neighbor ke andr, k-nearest neighbor algorithm milegi.

#     here we will use -> K-Nearest Neighbor classifier.

# Why we are using here classifier.
# bcoz the output of the dataset, means jo hm result nikalne wale hai,
# that are classification in nature. mtlb hmara answer 0 and 1 ke tor pe dekhne ko mil rha hai.
    
# that's why we are using here classification.
```

```python
knn = KNeighborsClassifier(n_neighbors = 3)
knn.fit(x_train,y_train)
```

```python
# # Hyperparameters of KNN

# knn = KNeighborsClassifier(shift+tab)


# ------------------------------------------------------------------------------

# 1. n_neighbors = 5
# KNeighborsClassifier(
#     n_neighbors=5,
#     *,

# 1st hyperparameter is n number  of neighbors.
#     means hm kitne neighbors bnane chahte hai.

# Default Neighbor is 5.

# We can change this neighbor according to our dataset.
# jaise jaise hmara dataset change hota rhega.
# waise waise hm, is neighbor ko change kr skte hai.
# ye check kr skte hai, ki kon sa neighbor hmare dataset ke liye best rhega.

# ----------------------------------------------------------------------------------

# 2. Which distance formula, we have to use?
# Manhattan
# Euclidian

# documentation me hi thoda niche jayenge.
#     to wha hme milega, p ki value


# p=1 manhattan 
# p=2 euclidian (Default)

# so, abhi hm by default wala distance hi use kr rhe hai euclidian.
#     but we can convert it into manhattan distance by changing p value.


# so, yha pe hme 2 cheeze dekhne hai.
#     1. no. of neighbors
#     2. distance formula
```

```python
knn.score(x_test,y_test)*100
```

```python
# n = 5, 92.5
# n = 10, 92.5
# n = 7, 93.75 (yes, improvement in accuracy by changing neighbor value)
```

### Applying loop to get neighbors best value

```python
# so, Now hm ek loop lgayenge, aur loop lgane ke badh me hm apne model ko ache treke se train krenge.

# hme yha pe neighbor ki right value lani pdegi, jiski bje se hmare model ki jo accuracy hai.
#     wo achi aye. aur mera model overfitting ke scenario me na chle jaye.
```

```python
for i in range(1,30):
    knn1 = KNeighborsClassifier(n_neighbors=i)
    knn1.fit(x_train,y_train)
    print(i,knn1.score(x_train,y_train)*100,knn1.score(x_test,y_test)*100)
```

```python
## Explanation
# -------------------

# Here, we have take range till 30
# //yha loop me model train ho rha hai.

# 1st, 2nd overfitting scenario.
# 3rd model almost shi hai.
# 6 best fitting.
#     and then niche ache answers dekhne ko nhi mil rhe hai.

# To shi neighbor ki value ki bat kre to, jo perfect neighbour nikl ke aya 
# hmare samne that is a 3.
# It means hme no. of neighbor ki value 3 krni pdegi.


# AT 3.
#     training and testing accuracy kafi ache aye hai
# aur jada difference bhi nhi hai.
#         it means hmara model overfitting ke scenario se ht chuka hai.
# It means hmara jo model hai, wo perfectly work krne wala hai.
```

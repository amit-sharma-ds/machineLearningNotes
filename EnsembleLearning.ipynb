```python
# Separating x and y axis data/ input, output / dependent & independent variable

x_a = dataset.iloc[:,:-1]
y_a = dataset["y"]

# //
# note: x axis ka data 2d me hona chahiye.
# ab 2d me kaise kre. we have already discuss in SLR
    # 1. agr more than 1 column hai, to wo automatically 2d me hi rhega.
    # 2. if iloc se separater kr rhe hai x ko, to wo 2d ho jayega.
    # 3. use double sq bracket: if dataset[["x"]] ese le rhe hai to.
```

```python
# splitting into train & test

from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_a,y_a,test_size=0.2,random_state=42)
```

```python
### Building the Model

# now we will use different different model
# through which we will train our dataset and check the model accuracy.

# If the accuracy will not good.
# Then we will use here "VOTING CLASSIFIER" means ensemble learning to improve the model accuracy.
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
# DT is a tree based algorithm
```

```python
# Now we will check the models accuracy
```

```python
dt = DecisionTreeClassifier() # calling the model/ object of model
dt.fit(x_train,y_train)
dt.score(x_train,y_train)*100 , dt.score(x_test,y_test)*100
```

```python
# //Note:

# Note: 1
    # ye hmne dataset ka sample liya hai.
    # isliye jitne bar code ko run krenge utne bar output different ayega.
    # bcoz sample rows ko khi se bhi le skta hai
    # (100.0, 92.0)
    
# Note 2: 
    # The model is scoring 100% on the training data, which indicates overfitting. This means the model is perfectly fitting the training data but losing some performance on the test data. To prevent overfitting:
    
    # Pruning: Trim the tree.
    # Cross-validation: Test the model on different subsets.
    # Regularization: Limit the tree depth.
    # Ensemble methods: Use Random Forest or Gradient Boosting.
    # These steps will help improve the model.
```

```python
sv = SVC()
sv.fit(x_train,y_train)
sv.score(x_train,y_train)*100 , sv.score(x_test,y_test)*100 

# svc ne kafi acha model bna ke diya hai.
# 96 & 96 without overfitting.

# but hme apne accuracy aur ache chahiye
```

```python
gnb = GaussianNB()
gnb.fit(x_train,y_train)
gnb.score(x_train,y_train)*100 ,  gnb.score(x_test,y_test)*100

# this is also best fit accuracy
```

```python
# Observation:

# sbse ache model svc ki aa rhe hai.
# bcoz its accuracy is high.
# DT (100.0, 94.5), SVC (97.125, 96.5), gnb (87.75, 88.5)

# but we want more good model as campare to above.
# which gives use answer with good accuracy

# For it We can use here "VOTING CLASSIFIER". means 
# WE CAN USE ENSEMBLE LEARNING.
```

```python
## Now, we will understand that how ensemble learning work.
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
## Creating estimators
# --------------------------------

# estimators : list of (str, estimator) tuples

# estimator should be list of tuples.
#     i. model name  ii. estimator name

li = [("dt1",DecisionTreeClassifier()),("sv1",SVC()),("gnb1",GaussianNB())]
```

```python
vc = VotingClassifier(li)
vc.fit(x_train,y_train)
```

```python
# ## Explanation
# # -------------------

# * vc object create kiya hmen. and VotingClassifier ko call.

# ---------------------------------------------------------------------
    
# * shift tab ->  vc = VotingClassifier()

#     Init signature:
#     VotingClassifier(
#         estimators,
#         *,
#         voting='hard',
#         weights=None,
#         n_jobs=None,
#         flatten_transform=True,
#         verbose=False,
#     )
#     Parameters
#     ----------
#     estimators : list of (str, estimator) tuples
    

# --------------------------------------------------------------------
    
# * voting classifier ko jb bhi call krte hai. means iske documentation me shift tab se jo ata hai.
# to voting classifier ko jb bhi hm call krte hai. to sbse pehle ata hai hmara estimator.
    
# * estimator: estimator means hm kon kon se models yha pe choose krne wale hai. wo yha pe dena hai.

# * weightage: iske andr hm, alg alg weightage ko define kr skte hai. joki hmne dekha tha ki, weightage average voting kaise kam krti hai.

# * Now, ye jo estimator hai, ye hme kaise dena hai.
# estimator hme list of tuple dena hai, is tuple me hme sbse pehle us model ka name likhna hai,
# and then estimator ka name dena hai, to is treke se ye hme pass krana hai.
```

```python
## checking score
vc.score(x_train,y_train)*100 , vc.score(x_test,y_test)*100

# this is training and testing score/ accuracy

# This is very good accuracy -> (97.625, 96.0)
# // and overfit bhi ni hai, bcoz training and testing accuracy are very close.
```

## Bagging

<span style="color:red">Bagging</span> (or Boostrap Aggregating) technique uses these subsets (bags) to get a fair
idea of the distribution (complete set). The size of subsets create for bagging may be
less than the original set.


img at 7.46.32

```python
## Explanation

i. Bagging Full name Boostrap Aggregating

ii. This is a parallel technique to improve our model.
```

```python
iii. How It works?

ye jo bhi hmare pass original Dataset hai, usko ye chote chote parts me todta hai,
                                                ---------------------
                                                        \|/
                                         This small small parts selected randomly


                   org. Data
   ---------------------------------------------
   |         |         |          |           |
   s1        s2        s3         s4          s5


This small small parts are selecting randomly. So, it may be possible that
ek subset ka data, dsure subset me bhi ho skta hai.
bcoz we are selecting subsets randomly.


For ex:             10000   <- we have 10000 rows
                      | 
                    80000 training    <- original data
             /    /   |   \   \
            s1   s2   s3  s4  s5

we have 10000 rows, 

and 10000 rows me se hmne training aur testing ke andr tod chuke hai, yha pr.
    
Training aur testing ke andr todne ke badh me, what we are doing here, is
8k rows hmare training ke andr aa gyi.

Now, ye jo 8k rows hai, isko hm alg alg sub sets ke andr todte hai, 


Remember: 
here subsets ko todte time, hm ek cheez dhyan me rkhte hai ki,
subsets should not be comparitively equal to original data.
    here org. data = is our training data which is here 8k.

mtlb subset hmesa original dataset less than hi hona chahiye.
ye cheez hme specially hmesha dhyan rkhna hai,
training data se km hi break down krna hai.
    
   +-------------------------------+
   | subset should be <  org. data |
   +-------------------------------+


No. of subsets depend on himself.
```

```python
In short iii) How to select no. of subsets

    10000 <-- dataset
      |
    80000  <-- training/ org. data
--------------
|  |  |  |  |
s1 s2 s3 s4 s5

subsets < org. data

No. of subsets depend on himself.
The parts/ subsets selected randomly.
```

### After subsets -> next process -> Bagggin Process

```python
alg alg subsets ready krne ke badh me next process start hota hai. --> Bagging Process
```

```python
Bagging Process
------------------

jitne bhi subsets hmne bnaye. 
lets' suppose we have ready 5 subsets. 
                   |
In 5 subsets ko, hm 5 models ko dete hai. 

And, these 5 models are the same model. mtlb same models ho skte hai. like I used DT here.

DT: 
subset 1: is decision tree ko hmne subset no. 1 dete hai. 
    aur subset no. 1 se training krte hai.
        aur training krne ke badh final prediction bna dete hai.

Similary: isi dt ko hmne subset 2 diya, isi dt ko hmne subset no. 3 diya.
    isi dt ko hmne subset no. 4 diya, isi ko 5 diya.

to hmne 5 alg alg decision tree bnaye.
        means  dt1, dt2, dt3, dt4, dt5
        hmne 5 alg alg models yha pe tyar kr liye. 
jo ki 5 alg alg subsets ke upr bne hai.
```

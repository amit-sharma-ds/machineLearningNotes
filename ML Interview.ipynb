```python
# 3. Encoding

# Because the machine learning algorithm not understand the categorical data.
# they only understand the numerical data.
# that's why we convert the categorical data into numerical is called encoding.


## Techniques

# Label Encoding – Converts categories into numeric labels (0, 1, 2, …).
# 🔹 Use when: The feature is ordinal (e.g., “Low” < “Medium” < “High”).

# One-Hot Encoding – Creates binary columns for each category (0/1).
# 🔹 Use when: The feature is nominal (no order), like “Red”, “Blue”, “Green”.

# Ordinal Encoding – Maps categories to integers reflecting their rank or order.
# 🔹 Use when: You have ordered categories, similar to Label Encoding but with explicit order awareness.

# Frequency Encoding – Replaces each category with its frequency or proportion in the data.
# 🔹 Use when: You have high-cardinality categorical features and want a compact numeric encoding.

# Target Encoding – Replaces categories with the mean of the target variable for that category.
# 🔹 Use when: You have many unique categories and sufficient data to reduce overfitting risk.
```

# Confusion Matrix & Classification Metrics

## Confusion Matrix
A table to evaluate a classification model by comparing predicted vs actual labels.

|                 | Predicted Positive | Predicted Negative |
|-----------------|-----------------|-----------------|
| **Actual Positive** | True Positive (TP)  | False Negative (FN) |
| **Actual Negative** | False Positive (FP) | True Negative (TN) |

---

## Key Metrics

1. **Accuracy**  
   $$ 
   Accuracy = \frac{TP + TN}{TP + TN + FP + FN} 
   $$  
   - Overall correctness of the model.

2. **Precision**  
   $$ 
   Precision = \frac{TP}{TP + FP} 
   $$  
   - How many predicted positives are actually positive.

3. **Recall (Sensitivity / True Positive Rate)**  
   $$ 
   Recall = \frac{TP}{TP + FN} 
   $$  
   - How many actual positives are correctly predicted.

4. **F1-Score**  
   $$ 
   F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall} 
   $$  
   - Harmonic mean of Precision and Recall; balances both.


# Unsupervised Learning

**Definition:**  
A type of machine learning where the model learns patterns or structure from data **without labeled outputs**.

---

## Types and Algorithms

### 1. Clustering
Groups similar data points together.  

**Algorithms:**  
- K-Means  
- Hierarchical Clustering  
- DBSCAN  

### 2. Dimensionality Reduction
Reduces the number of features while retaining important information.  

**Algorithms:**  
- Principal Component Analysis (PCA)  
- t-SNE (t-distributed Stochastic Neighbor Embedding)  
- UMAP  

### 3. Association Rule Learning
Finds interesting relationships or patterns between variables.  

**Algorithms:**  
- Apriori  
- Eclat  
- FP-Growth  

### 4. Anomaly Detection (Optional Sub-type)
Identifies rare or unusual data points.  

**Algorithms:**  
- Isolation Forest  
- One-Class SVM  
- Local Outlier Factor (LOF)


Underfitting: The model is too simple and cannot capture patterns in the data → both accuracies low, bias high, variance low.

Overfitting: The model is too complex and memorizes the training data → training accuracy high, testing accuracy low, bias low, variance high.

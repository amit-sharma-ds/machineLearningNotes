<u>//How to Identify dataset is polynomial</u>


* In SLR and MLR the dataset is following the simple linearity. means in graph in straight line.

* "Level Vs Salary.": when data is appear in this type of formation.

* But hme ek bar sure ho jana chahiye ki hmare dataset is following the polynomial feature or not.

<u>In short:</u>
* If the graph is not following the linearity the we use polynomial regression.

___
**<u>POLYNOMIAL REGRESSION</u>**
    
Polynomial Regression is a regression algorithm 
that models the relationship between a dependent(y) and independent variable(x) as
nth degree polynomial.

$$
Y = b_0 + b_1x_1 + b_2x_1^2 + b_3x_1^3 + \cdots + b_nx_1^n
$$


```python
# polynomial regression ke andr bhi independent and dependent variable same hi rehte hai.
# but jb hm graph ka nature dekhte hai to wo kuch is treke ka dikhai deta as.
# as show in above picture.

# so if the nature of graph is looks like this.
# generally graph ka jaise above image me hai usse different bhi ho skta hai.
# to jb hmare pass graph ke natures ese hote hai,
# then we can not apply simple linear regression.
```

### why we can not use SLR if the graph is in polynomial nature?

image at 5.40.32


```python
# see image at 5.40.32 above.

###### In image 1.

# when we apply SLR.  then our SLR line kuch is treh se pass hote hai (see image 1).
# and jb ye line is treh se pass hote hai in polynomial graph.  
# to ye ek underfitting line ho jata hai.


# Underfitting line means
# y org which is actual value
# y^ which is predict predicted value.
# dono ke bich bhut jada difference hai. which leads to more error.

# to jb hm esa graph wale data me SLR ka use krte hai to.
# SLR ki line straight hote hai. to wo sare data points ko ache se capture ni kr pata.

# jisse error jada ho jata hai means difference is more.
# error: actural yorg - predict value y^ 
#  to agr hme esa dataset me SLR y=mx+c ko apply kre to hmara model underfitting ke andr chala jata hai.


######### image 2
# to is cheez se bachne ke liye hm polynomial regression ko use krte hai.

# polymial reg. me hmare line ki eq change ho jate hai.
# and change hoke polynomial eq ke andr convert ho jata hai.

# image 2 me jo curvature hai, us curvature ke according hmare predicted line bnte hai.

# isme agr hme koi output check kre.
# actual ka and predicted line se bhi.
# to_dono ki line y ke corresponding same hi bnega.


# to polynomial curvature actual and predicted ka output same deta hai.


# In short.

# agr hm slr ka use ek polynomial dataset me use krege to
# hmara model underfit ho jayega.
# bcoz slr ek straight best fit line bnanta hai. jisse wo polynomial dataset ko capture ni kr pata.
```

### Eq of Polynomial regression

<u>Polynomial regression equation vs MLR: </u>
$$
Y = b_0 + b_1x_1 + b_2x_1^2 + b_3x_1^3 + \cdots + b_nx_1^n
$$

MLR eq: y = b0 + b1x1 + b2x2 + b3x3...
    
    
* b0 means constant.
* polynomial regression me hmare data ke andr power add ho jata hai,
* when we use power then our prediction line changes to polynomial line from the straight line.
(see in image at 5.41.26 below)


<u>Polynomial line</u>

* binomian, multinomial
* polynomial ke andr hmare line binomial bhi hota hai, multinomial bhi hota hai.

<u>How to decide binomial and multinomial?</u>

* so basically jb tk hm power 2 ki eq ko use krte hai. then that is binomial eq.
* when we use power more than 2 then our polynomial eq also increases.

## COST FUNCTION

* A cost function is an important parameter that determines how well a machine
learning model performs for a given dataset.

* Cost function is a measure of how wrong the model is in estimating the relationship
between X(input) and Y(output) Parameter.

## TYPES OF COST FUNCTION

1. Regression Cost Function
2. Classification Cost Function

```python
# //
###### explanation above

# cost function are divided into 2 types.
# reg.cf and class. cf

# Regression CF: If the output is in continuous in nature, then we apply Reg. CF.
# classification: If the output is in classification in nature then we apply Classificatiion CF.
#     Classification means True False, Yes No, Cat and Dog etc. 
```

**<span style="color:red"><u>1. Regression Cost Function:</u></span>**

Regression models are used to make predictions for continuous variables.

- **MSE** (Mean Square Error)  
- **RMSE** (Root Mean Square Error)  
- **MAE** (Mean Absolute Error)  
- **R<sup>2</sup>** Accuracy  

**<span style="color:red"><u>2. Binary Classification Cost Functions:</u></span>**
    
Classification models are used to make predictions of 
categorical variables, such as predictions for 0 or 1,Cat 
or dog, etc.
    
**<span style="color:red"><u>3. Multi-class Classification Cost Functions:</u></span>**
    
A multi-class classification cost function is used in the 
classification problems for which instances are 
allocated to one of more than two classes.

- **Binary Cross Entropy Cost Function or Log Loss Function**
    
    

```python
# # Explation above

# 1) Regression CF: 4

# #### Classification cost function are divided into 2 parts.

# 1) Binary Classification CF: where our output is divided into are divided into 
#     (True, False), (Yes, NO), or (Cat, dog) etc.
    
# 2) Multi-class Classification CF:
# where our output is more than 2.
# like (cat, dog, cow, buffalow) and apple, banana, carrot.



# ** cost function like binary cross entrophy and log loss function,
# helps in better separation in our data or can do better prediction.
```

```python
#### Understanding alpha hyper parameter in Lasso Regression:
# ----------------------------------------------------------------
# Alpha is a regularization parameter in Lasso regression that 
# controls the strength of penalty applied to the model's 
# coefficients to prevent overfitting.

# Alpha in Lasso regression controls the regularization strength.
    # * Small alpha (e.g., 0.01): Less regularization, the model may become more complex.
    # * Large alpha (e.g., 1.0): More regularization, the model becomes simpler, and some coefficients may shrink to zero.
# alpha=0.5 means medium regularization strength.

# // In short: it ask penalty

# Note: alpha is a hyperparameter bcoz
# A hyperparameter is a parameter whose value is set before the learning process begins (not learned from the data).

#### Penalty:
# --------------------
# Here, penalty means adding a constraint (L1 regularization) 
# to the model's loss function to shrink or reduce the coefficients 
# of less important features, often pushing some of them to zero, 
# which helps in simplifying the model and preventing overfitting. 
```

# Decision Tree

* Decision Tree is a Supervised learning technique that can be used for both 
classification and Regression problems, but mostly it is preferred for solving 
Classification problems.

* In order to build a tree, we use the CART algorithm, which stands for Classification
and Regression Tree algorithm.

```python
# DT Explanation

# It works in both type of data (regression and classification).
# chahe hmara data regression hai, chahe hmara data classification analysis ko follow krta hai.
    
# The algorithm used in it is CART algorith.
#     CART (Classification and Regression Tree algorithm).

# Ye algorithm hmare data ko splitting kr dete hai.
#     and konsi splitting krte hai.  --> Binary splitting.

# Means same as conditional statement ke treh hmare data ko splitting krte hai,
# aur jo bhi hme prediction krna hai, wo predictions answer hme likh ke dete hai.


# for example: see below image

# DT is treh ke data ko conditional statements me break krta hai,
# mtlb binary splitting krta hai, aur splitting krke hmara jo answer hai,
# jo bhi prediction hm isse krwana chahte hai, wo prediction hme lake dete hai.
#     That is a CART algorith.
```

## Important Terminology related to Decision Trees

* <span style="color:red"><u>Root Node:</u></span> It represents the entire population or sample and this further gets divided
into two or more homogeneous sets.

* <span style="color:red"><u>Splitting:</u></span> It is a process of dividing a node into two or more sub-nodes.

* <span style="color:red"><u>Decision Node:</u></span> When a sub-node splits into further sub-nodes, then it is called the
decision node.

* <span style="color:red"><u>Leaf / Terminal Node:</u></span> Nodes do not split is called Leaf or Terminal node.

* <span style="color:red"><u>Pruning:</u></span> When we remove sub-nodes of a decision node, this process is called pruning.
    You can say the opposite process of splitting.

* <span style="color:red"><u>Brach / Sub-Tree:</u></span> A subsection of the entire tree is called branch or sub-tree.

* <span style="color:red"><u>Parent and Child Node:</u></span> A node, which is divided into sub-nodes is called a parent node
    of sub-nodes whereas sub-nodes are the child of a parent node.

```python
# ## Explanation terminology

# Root Node: Starting node is root node.

# splitting: Means todna. we split the decision node into 2 parts. means sub-nodes ke andr
# todte hai, that is called of splitting.

# Decision Node: Nodes jo splitting hone ke badh bnte hai.

# Leaf Node: Also called Terminal Node. means the last node.

# Pruning: Kafi bar hmare decision tree kafi lamba bn jata hai,
# jiske bje se over fitting ke chances bdh jate hai,
# or is over fitting se bchne ke liye hm pruning krte hai.

# Basic simple means of pruning is cut krna.

#     For ex: jb tree ke braches bhut jada bdh jate hai,
# to hm us tree ke braches ko cut kr dete hai,
# usi treke se ye jo hmara dt hai, ye kafi complex ho jata hai,
# jiske bje se ye overfitting me chlta jata hai,
# us overfitting ke scenario ko km krne ke liye.
#     hm yha pe pruning ka use krte hai.

# Pruning se hme jitna chahiye, hm uske hisab se dt ko cut kr dete hai.
#     that is called of pruning.


# Brach/ Sub-Tree: means kisi bhi decision node ko hm further splitting krte rhenge.
#         to brach or sub tree bnte rhenge.

# Parent and Child Node: 
#     jo hmara decision nodes hota hai, uske niche ke nodes child node.
#         aur jo decision nodes hota hai, that is called of parent node.
```

## ATTRIBUTE SELECTION MEASURES

**This measurement, we can easily select the best attribute for the nodes 
of the tree. There are two popular techniques for ASM, which are:**

* Information Gain
* Entropy / Gini Index

```python
# ENTROPY AND GINI INDEX ke help se hi hm pehla Root Node kon sa hoga.
#     aur uske badh uska next decision nodes kon sa hone wala hai.
#     wo pta lgate hai. Let's understand how.
```

* <span style="color:red">Entropy:</span> Entropy is a metric to measure the impurity in a given attribute. It 
specifies randomness in data.

    $$
Entropy(S) = - P(yes) \log_2 P(yes) - P(no) \log_2 P(no)
$$


Where:

S= Total number of samples

P(yes) = probability of yes

P(no) = probability of no

```python
# Entropy means impurity. 
# means jo hmare pass sbse low impure data hoga.
#     Then wo hmara decision nodes bnne wala hai. Let's understand how.
```

* <span style="color:red">Information Gain:</span> Information gain is the measurement of changes in entropy 
after the segmentation of a dataset based on an attribute. It calculates how 
much information a feature provides us about a class.

    Information Gain = Entropy(S) - [(Weighted Avg) * Entropy(each feature)]
<br>

* <span style="color:red">Entropy:</span> Entropy is a metric to measure the impurity in a given attribute. It 
specifies randomness in data.

    Entropy(S) = - P(yes) * log2 P(yes) - P(no) * log2 P(no)

Where:
S = Total number of samples
P(yes) = probability of yes
P(no) = probability of no

<br>
    
* <span style="color:red">Gini Index:</span> Gini index is a measure of impurity or purity used while creating a 
decision tree in the CART(Classification and Regression Tree) algorithm.

* An attribute with the low Gini index should be preferred as compared to the high 
Gini index. 

    Gini Index = 1 - Σ (Pj)²

### training aur testing me kitne difference pe overfitting khenge

General Rule (simple thumb rule)

Agar Training accuracy – Testing accuracy > 5% (0.05)
to Overfitting consider karte hai.

Example:

Training = 93.77%

Testing = 85.56%

Difference = 93.77 − 85.56 = 8.21% (8% se zyada)

So ye overfitting hai.
